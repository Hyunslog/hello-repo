주석으로 처리된 코드는 실행이 되는지 테스트해보는 예제입니다.
코드를 실행해보면 주석 처리된 코드는 출력 되지 않고 주석 처리가 되지않은 부분은 출력된 것을 확인 해볼 수 있니다.

````python
price = int(input("상품 가격(원): "))   # 입력: 숫자를 받아 price에 저장
total = price * 1.1                  # 처리: 부가세 10%를 더한 금액 계산
if(total>11000):
    print("상품 가격이 초과했습니다.")
else:
    print("결제 금액:", int(total), "원")   # 출력: 소수점을 버리고 화면에 표시
````

다음 예제는 input()을 통해 코드 실행 후, 입력받은 가격을 price 변수에 저장하고, 이 가격에 10% 부가세가 추가된 합계를 출력하는 예제입니다. 저는 여기에 추가로 부가세 합계 금액이 11,000원이 넘어가면 "삼품 가격 초과" 되었다는 출력을 하도록 하는 조건문을 추가적으로 테스트 해봤습니다.
다음 input 값으로 11,000원을 입력한 결과 총 금액이 11000*1.1=12100 되므로, 조건문이 정상 실행 되어 상품 가격이 초과되었다는 출력 결과를 확인할 수 있었습니다.

````python
print("안녕하세요, 청년식자재 마트입니다.")
print("다음은 저희 메뉴 입니다.")

print("="*20)
print(" "*8 + "메뉴" + " "*8)
print("="*20)
onion_price = int(input("양파 가격(원): "))
tomato_price = int(input("토마토 가격(원): "))
potato_price = int(input("감자 가격(원): "))

print("="*20)

onion_count = int(input("구입하실 양파의 수량을 입력 해주세요 : "))
tomato_count = int(input("구입하실 토마토의 수량을 입력 해주세요 : "))
potato_count = int(input("구입하실 감자의 수량을 입력 해주세요 : "))

onion_sum=onion_price*onion_count
tomato_sum=tomato_price*tomato_count
potato_sum=potato_price*potato_count

sum = onion_sum+tomato_sum+potato_sum
tax = int(sum * 0.1)
total_price = sum+tax

print("-"*20)
print("선택하신 메뉴는 다음과 같습니다.")

print("-"*20)
print(f"양파 / 수량: {onion_count}개, 가격: ", onion_sum, "원")
print(f"토마토 / 수량: {onion_count}개, 가격: ", tomato_sum, "원")
print(f"감자 / 수량: {onion_count}개, 가격: ", potato_sum, "원")
print("-"*20)
print("합계:", sum, "원")
print("세액:", tax, "원")
print("="*20)
print("결제 금액:", total_price, "원")
````

이번 예제는 코드 실행시, input 박스를 통해 양파, 토마토, 감자의 가격과 수량을 입력 받아 품목 개당 가격 * 수량을 계산한 뒤, 모든 물건의 가격 합계와 세액 10프로를 계산하여 결제할 금액을 출력하는 실습이었습니다.

````python
potato = int(input("감자의 상품 가격(원): "))
totato = int(input("토마토의 상품 가격(원): "))
onion = int(input("양파의 상품 가격(원): "))

potato_price = potato * 3
tomato_price = totato * 3
onion_price = onion * 3
````

````python
print("결제 금액:", int(potato_price), "원")
````

````python
print("결제 금액:", int(tomato_price), "원")
````

````python
print("결제 금액:", int(onion_price), "원")
````

마지막 예제는 파이썬의 코드가 상단에서부터 순차적으로 실행된다는 것을 체크해 볼 수 있는 실습이었습니다.
onion_price가 처음에는 입력되어있지 않않고, ```print("결제 금액:", int(onion_price), "원")```코드 실행 결과에 오류가 발생 하는 것을 확인할 수 있었습니다. 따라서 ```onion_price```의 값을 출력하기 위해서는 상단에 먼저 변수로 선언되어 있어야 출력이 가능하다는 것을 이 실습을 통해 확인할 수 있었습니다.
