```python
price = int(input("상품 가격(원): "))   # 입력: 숫자를 받아 price에 저장
total = price * 1.1                  # 처리: 부가세 10%를 더한 금액 계산
if(total>11000):
    print("상품 가격이 초과했습니다.")
else:
    print("결제 금액:", int(total), "원")   # 출력: 소수점을 버리고 화면에 표시
```

```python
print("안녕하세요, 청년식자재 마트입니다.")
print("다음은 저희 메뉴 입니다.")

print("-"*20)
price_1 = int(input("양파 가격(원): "))
price_2 = int(input("토마토 가격(원): "))
price_3 = int(input("감자 가격(원): "))

sum = price_1+price_2+price_3
tax = int(sum * 0.1)
total_price = sum+tax

print("-"*20)
print("선택하신 메뉴는 다음과 같습니다.")
print("-"*20)
print("양파 1개, 가격: ", price_1, "원")
print("토마토 1개, 가격: ", price_2, "원")
print("감자 1개, 가격: ", price_3, "원")
print("-"*20)
print("합계:", sum, "원")
print("세액:", tax, "원")
print("="*20)
print("결제 금액:", total_price, "원")
```
