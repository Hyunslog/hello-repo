```python
import random
import time

class Character:
    def __init__(self, name, job):
        self.name = name
        self.job = job
        self.level = 1
        self.max_hp = 100
        self.hp = 100
        self.max_mp = 50
        self.mp = 50
        self.gold = 100
        self.exp = 0
        self.max_exp = 100

        # 직업별 능력치 보정
        if job == '전사':
            self.max_hp += 50
            self.hp = self.max_hp
            self.atk = 15
            self.spell_power = 5
        elif job == '마법사':
            self.max_mp += 50
            self.mp = self.max_mp
            self.atk = 8
            self.spell_power = 20
        else: # 도적 등 기본값
            self.atk = 12
            self.spell_power = 10

    def show_status(self):
        print(f"\n=== {self.name} ({self.job}) 상태 ===")
        print(f"레벨: {self.level} | 경험치: {self.exp}/{self.max_exp}")
        print(f"체력: {self.hp}/{self.max_hp} | 마력: {self.mp}/{self.max_mp}")
        print(f"공격력: {self.atk} | 주문력: {self.spell_power}")
        print(f"소지금: {self.gold} 골드")
        print("===============================\n")

    def check_level_up(self):
        while self.exp >= self.max_exp:
            self.exp -= self.max_exp
            self.level += 1
            # 레벨이 오를수록 필요 경험치 통이 커짐 (이전 필요 경험치의 1.5배)
            self.max_exp = int(self.max_exp * 1.5)

            # 능력치 상승
            self.max_hp += 20
            self.max_mp += 10
            self.hp = self.max_hp
            self.mp = self.max_mp
            self.atk += 3
            self.spell_power += 3
            print(f"\n★ 축하합니다! 레벨업 하였습니다! (현재 레벨: {self.level}) ★")
            self.show_status()

class Monster:
    def __init__(self, player_level):
        # 플레이어 레벨 부근의 몬스터 레벨 생성
        self.level = max(1, player_level + random.randint(-1, 2))
        names = ['슬라임', '고블린', '오크', '스켈레톤', '드래곤']
        # 레벨에 맞는 몬스터 이름 선택
        name_idx = min(self.level - 1, len(names) - 1)
        self.name = f"LV.{self.level} {names[name_idx]}"

        # 능력치 결정 (레벨 비례)
        self.max_hp = 40 + (self.level * 15)
        self.hp = self.max_hp
        self.atk = 5 + (self.level * 3)

        # 처치 시 획득 보상 (몬스터 레벨이 높을수록 많은 경험치와 골드)
        self.exp_reward = 20 + (self.level * 15)
        self.gold_reward = 10 + (self.level * 8)

def battle(player):
    monster = Monster(player.level)
    print(f"\n앗! 야생의 {monster.name}(이)가 나타났다!")

    while player.hp > 0 and monster.hp > 0:
        print(f"\n[{player.name} HP:{player.hp}/{player.max_hp} MP:{player.mp}/{player.max_mp}] VS [{monster.name} HP:{monster.hp}/{monster.max_hp}]")
        print("1. 일반 공격 | 2. 스킬 사용 (MP 15) | 3. 도망치기")
        choice = input("행동을 선택하세요: ")

        # 플레이어 턴
        if choice == '1':
            damage = random.randint(int(player.atk*0.8), int(player.atk*1.2))
            monster.hp -= damage
            print(f"\n{player.name}의 물리 공격! {monster.name}에게 {damage}의 피해를 주었습니다.")
        elif choice == '2':
            if player.mp >= 15:
                player.mp -= 15
                damage = random.randint(int(player.spell_power*1.2), int(player.spell_power*1.8))
                monster.hp -= damage
                print(f"\n{player.name}의 스킬 발동! {monster.name}에게 {damage}의 강력한 마법 피해를 주었습니다.")
            else:
                print("\n마력이 부족합니다! 일반 공격으로 전환합니다.")
                damage = random.randint(int(player.atk*0.8), int(player.atk*1.2))
                monster.hp -= damage
                print(f"{player.name}의 물리 공격! {monster.name}에게 {damage}의 피해를 주었습니다.")
        elif choice == '3':
            if random.random() < 0.5:
                print("\n무사히 도망쳤습니다!")
                return
            else:
                print("\n도망치는데 실패했습니다!")
        else:
            print("\n잘못된 선택입니다. 턴이 넘어갑니다.")

        if monster.hp <= 0:
            break

        time.sleep(0.5)

        # 몬스터 턴
        monster_damage = random.randint(int(monster.atk*0.8), int(monster.atk*1.2))
        player.hp -= monster_damage
        print(f"{monster.name}의 반격! {player.name}에게 {monster_damage}의 피해를 주었습니다.")

        time.sleep(0.5)

    if player.hp <= 0:
        print(f"\n{player.name}(이)가 쓰러졌습니다... 게임 오버!")
        return False
    else:
        print(f"\n승리했습니다! {monster.name}을 처치했습니다.")
        print(f"획득 보상: {monster.exp_reward} EXP, {monster.gold_reward} Gold")
        player.exp += monster.exp_reward
        player.gold += monster.gold_reward
        player.check_level_up()
        return True

def main():
    print("=== 간단한 파이썬 턴제 RPG ===")
    name = input("캐릭터 이름을 입력하세요: ")
    print("직업을 선택하세요:")
    print("1. 전사 (높은 체력, 안정적인 물리 공격)")
    print("2. 마법사 (낮은 체력, 강력한 마법 스킬)")
    job_choice = input("번호 선택: ")

    job = '전사' if job_choice == '1' else '마법사'
    player = Character(name, job)
    player.show_status()

    while True:
        print("\n무엇을 하시겠습니까?")
        print("1. 모험하기 (몬스터 탐색) | 2. 여관에서 쉬기 (50 Gold 필요, 체력/마력 회복) | 3. 상태 보기 | 4. 게임 종료")
        action = input("선택: ")

        if action == '1':
            alive = battle(player)
            if alive is False:
                print("게임을 다시 시작해 주세요.")
                break
        elif action == '2':
            if player.gold >= 50:
                player.gold -= 50
                player.hp = player.max_hp
                player.mp = player.max_mp
                print("\n여관에서 푹 쉬었습니다! 체력과 마력이 모두 회복되었습니다.")
            else:
                print("\n골드가 부족합니다!")
        elif action == '3':
            player.show_status()
        elif action == '4':
            print("\n게임을 종료합니다. 플레이해주셔서 감사합니다!")
            break

# 게임 실행
main()
```
